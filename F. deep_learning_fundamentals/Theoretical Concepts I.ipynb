**1. Generic Facts about Universal Approximation Theorem (UAT):**

UAT was first proved in 1989 for sigmoid activation functions.

It states that a feedforward neural network with at least one hidden layer can approximate any continuous function on a compact set.

The theorem shows the expressive power of NNs, not their training efficiency. It applies to a wide class of activation functions, including sigmoid, ReLU, and tanh.

UAT guarantees existence of weights, but not how to find them. It only applies to continuous functions, not arbitrary discontinuous ones.

Compact set means the function is bounded and closed in its domain. A neural network can approximate with arbitrary accuracy if allowed enough hidden units.

UAT is independent of optimization algorithms like gradient descent. The result is mathematical and does not imply practical efficiency.

It does not specify the size of the network required for approximation.

A single hidden layer is theoretically enough, but may require an exponential number of neurons.

Deeper architectures may approximate functions with fewer neurons than shallow ones. UAT is a foundation for why neural networks are powerful universal models.

It applies to both regression and classification tasks indirectly. The error bound can be made arbitrarily small, but not necessarily zero. UAT relates to the "expressive completeness" of neural nets.

In practice, optimization and generalization issues matter more than UAT guarantees. UAT is one of the main theoretical justifications for using neural networks widely.
