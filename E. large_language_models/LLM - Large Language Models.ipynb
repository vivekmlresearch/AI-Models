# # Stock Data Analysis and LLM Summary

```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from transformers import pipeline

# Function to calculate RSI
def calculate_rsi(data, window=14):
    delta = data['Close'].diff(1)
    gain = delta.where(delta > 0, 0)
    loss = -delta.where(delta < 0, 0)

    avg_gain = gain.rolling(window=window, min_periods=1).mean()
    avg_loss = loss.rolling(window=window, min_periods=1).mean()

    rs = avg_gain / avg_loss
    rsi = 100 - (100 / (1 + rs))
    return rsi

# Function to calculate VWAP
def calculate_vwap(data):
    vwap = (data['Close'] * data['Volume']).cumsum() / data['Volume'].cumsum()
    return vwap

# Function to perform live stock analysis
def live_stock_analysis(stock_symbol, model_name='distilgpt2'):
    # Fetch stock data from Yahoo Finance
    stock_data = yf.download(stock_symbol, period='5d', interval='1h')

    # Ensure there is enough data to proceed
    if stock_data.empty:
        print(f"No data found for {stock_symbol}. Please check the stock symbol or try again later.")
        return

    # Drop rows with NaN values to avoid indexing issues
    stock_data.dropna(inplace=True)
    if stock_data.empty:
        print(f"Not enough data after cleaning for {stock_symbol}. Cannot perform analysis.")
        return

    # Calculate RSI and VWAP
    stock_data['RSI'] = calculate_rsi(stock_data)
    stock_data['VWAP'] = calculate_vwap(stock_data)

    # Plotting the stock closing price, RSI, and VWAP
    fig, ax = plt.subplots(3, 1, figsize=(5, 5))

    ax[0].plot(stock_data.index, stock_data['Close'], label='Close Price', color='blue')
    ax[0].set_title(f'{stock_symbol} Close Price')
    ax[0].set_ylabel('Price')
    ax[0].legend()

    ax[1].plot(stock_data.index, stock_data['RSI'], label='RSI', color='red')
    ax[1].set_title('Relative Strength Index (RSI)')
    ax[1].set_ylabel('RSI')
    ax[1].axhline(70, color='gray', linestyle='--')
    ax[1].axhline(30, color='gray', linestyle='--')
    ax[1].legend()

    ax[2].plot(stock_data.index, stock_data['VWAP'], label='VWAP', color='green')
    ax[2].set_title('Volume Weighted Average Price (VWAP)')
    ax[2].set_ylabel('VWAP')
    ax[2].legend()

    plt.tight_layout()
    plt.show()

    # Prepare data for LLM
    if len(stock_data) > 0:
        analysis_text = f"The latest closing price of {stock_symbol} is {float(stock_data['Close'].iloc[-1]):.2f}. " \
                        f"The RSI is {float(stock_data['RSI'].iloc[-1]):.2f}, suggesting {'overbought' if stock_data['RSI'].iloc[-1] > 70 else 'oversold' if stock_data['RSI'].iloc[-1] < 30 else 'neutral'} conditions. " \
                        f"The VWAP is {float(stock_data['VWAP'].iloc[-1]):.2f}."
    else:
        print(f"Not enough data to generate analysis for {stock_symbol}")
        return

    # Load the LLM model
    llm_pipeline = pipeline('text-generation', model=model_name, trust_remote_code=True)

    # Generate analysis summary using LLM
    generated_analysis = llm_pipeline(analysis_text, max_length=100, truncation=True)[0]['generated_text']

    print("\nGenerated Stock Analysis Summary:\n")
    print(generated_analysis)

# Example usage
if __name__ == "__main__":
    live_stock_analysis('AAPL', model_name='distilgpt2')
```

# Weather Data Analysis and LLM Summary

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from transformers import pipeline
import random

# Function to generate sample weather data
def generate_weather_data(days=7, hours_per_day=24):
    date_range = pd.date_range(end=pd.Timestamp.now(), periods=days * hours_per_day, freq='H')
    data = {
        'Temperature': [random.uniform(15, 35) for _ in range(len(date_range))],  # Temperature in Celsius
        'Humidity': [random.uniform(30, 90) for _ in range(len(date_range))],      # Humidity in percentage
        'WindSpeed': [random.uniform(0, 20) for _ in range(len(date_range))]       # Wind speed in km/h
    }
    weather_data = pd.DataFrame(data, index=date_range)
    return weather_data

# Function to perform weather analysis
def weather_analysis(model_name='distilgpt2'):
    # Generate sample weather data
    weather_data = generate_weather_data()

    # Calculate key metrics: average temperature, humidity, wind speed
    avg_temperature = weather_data['Temperature'].mean()
    avg_humidity = weather_data['Humidity'].mean()
    avg_wind_speed = weather_data['WindSpeed'].mean()

    # Plotting the weather data
    fig, ax = plt.subplots(3, 1, figsize=(8, 6))

    ax[0].plot(weather_data.index, weather_data['Temperature'], label='Temperature', color='blue')
    ax[0].set_title('Temperature Over Time')
    ax[0].set_ylabel('Temperature (C)')
    ax[0].legend()

    ax[1].plot(weather_data.index, weather_data['Humidity'], label='Humidity', color='red')
    ax[1].set_title('Humidity Over Time')
    ax[1].set_ylabel('Humidity (%)')
    ax[1].legend()

    ax[2].plot(weather_data.index, weather_data['WindSpeed'], label='Wind Speed', color='green')
    ax[2].set_title('Wind Speed Over Time')
    ax[2].set_ylabel('Wind Speed (km/h)')
    ax[2].legend()

    plt.tight_layout()
    plt.show()

    # Prepare data for LLM
    analysis_text = f"The average temperature for the past week is {avg_temperature:.2f} degrees Celsius. " \
                    f"The average humidity is {avg_humidity:.2f}%, indicating {'high' if avg_humidity > 70 else 'moderate' if avg_humidity > 40 else 'low'} humidity levels. " \
                    f"The average wind speed is {avg_wind_speed:.2f} km/h."

    # Load the LLM model
    llm_pipeline = pipeline('text-generation', model=model_name, trust_remote_code=True)

    # Generate analysis summary using LLM
    generated_analysis = llm_pipeline(analysis_text, max_length=100, truncation=True)[0]['generated_text']

    print("\nGenerated Weather Analysis Summary:\n")
    print(generated_analysis)

# Example usage
if __name__ == "__main__":
    weather_analysis(model_name='distilgpt2')
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from transformers import pipeline
import kaggle
import os
import zipfile

# Download Kaggle dataset
kaggle.api.authenticate()

# Set up Kaggle dataset information
kaggle_dataset = 'new-york-city/nyc-taxi-trip-data'
dataset_zip_path = 'nyc_taxi_trip_data.zip'
dataset_folder = 'nyc_taxi_trip_data'

# Download the dataset
if not os.path.exists(dataset_zip_path):
    kaggle.api.dataset_download_files(kaggle_dataset, path='.', unzip=False)

# Unzip dataset if not already unzipped
if not os.path.exists(dataset_folder):
    with zipfile.ZipFile(dataset_zip_path, 'r') as zip_ref:
        zip_ref.extractall(dataset_folder)

# Load the retail sales data
retail_data = pd.read_csv(f'{dataset_folder}/yellow_tripdata_2019-01.csv')  # Modify the path if necessary

# Function to perform NYC taxi trip analysis
def nyc_taxi_trip_analysis(model_name='distilgpt2'):
    # Calculate key metrics: average sales amount, number of customers, number of transactions
    avg_trip_distance = retail_data['trip_distance'].mean()
    avg_fare_amount = retail_data['fare_amount'].mean()
    avg_trip_duration = (pd.to_datetime(retail_data['tpep_dropoff_datetime']) - pd.to_datetime(retail_data['tpep_pickup_datetime'])).mean().total_seconds() / 60

    # Plotting the retail data
    fig, ax = plt.subplots(3, 1, figsize=(10, 10))

    ax[0].hist(retail_data['trip_distance'], bins=50, label='Trip Distance', color='blue', alpha=0.7)
    ax[0].set_title('Trip Distance Distribution')
    ax[0].set_ylabel('Frequency')
    ax[0].legend()

    ax[1].hist(retail_data['fare_amount'], bins=50, label='Fare Amount', color='red', alpha=0.7)
    ax[1].set_title('Fare Amount Distribution')
    ax[1].set_ylabel('Frequency')
    ax[1].legend()

    ax[2].hist((pd.to_datetime(retail_data['tpep_dropoff_datetime']) - pd.to_datetime(retail_data['tpep_pickup_datetime'])).dt.total_seconds() / 60, bins=50, label='Trip Duration (minutes)', color='green', alpha=0.7)
    ax[2].set_title('Trip Duration Distribution')
    ax[2].set_ylabel('Frequency')
    ax[2].legend()

    plt.tight_layout()
    plt.show()

    # Prepare data for LLM
    analysis_text = f"The average trip distance is {avg_trip_distance:.2f} miles. " " " \
                    f"The average fare amount is ${avg_fare_amount:.2f}. " " " \
                    f"The average trip duration is {avg_trip_duration:.2f} minutes."""

    # Load the LLM model
    llm_pipeline = pipeline('text-generation', model=model_name, trust_remote_code=True)

    # Generate analysis summary using LLM
    generated_analysis = llm_pipeline(analysis_text, max_length=100, truncation=True)[0]['generated_text']

    print("\nGenerated Retail Sales Analysis Summary:\n")
    print(generated_analysis)

# Example usage
if __name__ == "__main__":
    nyc_taxi_trip_analysis(model_name='distilgpt2')
```

```python
%%bash
cat > /content/rag_demo.py <<'PY'
#!/usr/bin/env python3
"""
RAG Playground (pure Python, no external deps)
----------------------------------------------

This single-file script helps you understand four concepts:

1) LLM (Large Language Model) – we'll *simulate* one with simple templated summarization.
2) Vector Database – we'll build a tiny in-memory "vector store" and do cosine similarity search.
3) RAG (Retrieval-Augmented Generation) – retrieve relevant chunks, then generate an answer from them.
4) Reranking – after an initial dense retrieval, re-order results using a second scoring function (BM25).

Run it from a terminal:

    python rag_demo.py --query "What is RAG and how does reranking help?"

Or just run without arguments and you'll be prompted for a question.
The code is heavily commented—read along as you run it.
"""

from __future__ import annotations
import argparse
import math
import re
from collections import Counter
from dataclasses import dataclass
from typing import Dict, List, Tuple

# ----------------------------
# 0) Toy "knowledge base"
# ----------------------------

DOCS = [
    {
        "id": "doc_llm",
        "title": "Large Language Models (LLMs)",
        "text": (
            "Large Language Models, or LLMs, are neural networks trained on massive text corpora "
            "to predict the next token in a sequence. Modern LLMs are typically based on the "
            "Transformer architecture, which uses self-attention to model long-range dependencies. "
            "LLMs are capable of few-shot learning and instruction following. They do not have a persistent memory "
            "of facts unless such facts are included in the prompt, fine-tuning, or tools. "
            "At inference time, an LLM takes a prompt and generates text token by token, guided by probabilities."
        ),
    },
    {
        "id": "doc_vector_db",
        "title": "Vector Databases in a Nutshell",
        "text": (
            "A vector database stores high-dimensional numeric representations (embeddings) of data like text, images, or audio. "
            "It enables fast similarity search (nearest-neighbor queries) so you can find items 'close' to a query vector. "
            "Under the hood, vector databases often use specialized indexes such as HNSWgraphs or IVF-Flat to accelerate search. "
            "You typically compute embeddings with a model, insert them alongside your raw text, and query with another embedding."
        ),
    },
    {
        "id": "doc_rag",
        "title": "What is RAG? Retrieval-Augmented Generation",
        "text": (
            "Retrieval-Augmented Generation (RAG) improves LLM responses by injecting fresh, grounded context retrieved from a knowledge base. "
            "Typical steps are: chunk your documents, embed the chunks into vectors, store them in a vector database, retrieve top-k chunks for a user query, "
            "and provide those chunks to the LLM as context. RAG reduces hallucinations and lets models answer about data they were never trained on."
        ),
    },
    {
        "id": "doc_rerank",
        "title": "Reranking: From Recall to Precision",
        "text": (
            "Reranking is a second-stage ranking step after an initial retrieval. "
            "You might first retrieve candidates with a fast dense similarity search, then rerank a small set using a stronger but slower method. "
            "Common rerankers include BM25 lexical scoring or cross-encoders that score a (query, document) pair with a transformer. "
            "Reranking improves precision by pushing the most relevant passages to the top."
        ),
    },
    {
        "id": "doc_pipeline",
        "title": "Putting It Together: A Simple Pipeline",
        "text": (
            "A basic pipeline is: (1) User asks a question. (2) Convert query into an embedding and search the vector database to get candidate chunks. "
            "(3) Rerank the candidates using BM25 or a cross-encoder. (4) Send the best chunks to the LLM along with the question. (5) The LLM writes an answer "
            "and may cite the sources. This balances speed (fast retrieval) with accuracy (reranking and grounded context)."
        ),
    },
]

# ----------------------------
# 1) Simple tokenizer / stopwords
# ----------------------------

STOPWORDS = set(
    """
a an and are as at be but by for from has have if in into is it its of on or such that the their then there these they this to was were will with you your
""".split()
)

WORD_RE = re.compile(r"[a-z0-9']+")

def tokenize(text: str) -> List[str]:
    """Lowercase, split on words, drop stopwords."""
    toks = WORD_RE.findall(text.lower())
    return [t for t in toks if t not in STOPWORDS]

# ----------------------------
# 2) Tiny "Vector DB" (dense retrieval via cosine on term-frequency vectors)
# ----------------------------

@dataclass
class DocVector:
    doc_id: str
    title: str
    vector: Dict[str, float]  # term -> tf weight
    norm: float               # L2 norm for cosine
    length: int               # number of terms (for BM25)

class VectorDB:
    """
    Minimal in-memory vector store.
    - Builds a vocabulary from tokens.
    - Stores simple TF vectors (term frequency) for each document chunk.
    - Performs cosine similarity search.
    Also tracks document frequencies to support BM25 reranking.
    """
    def __init__(self):
        self._docs: Dict[str, DocVector] = {}
        self.doc_freq: Counter = Counter()
        self.N: int = 0  # number of docs
        self.avgdl: float = 0.0

    def add(self, doc_id: str, title: str, text: str):
        tokens = tokenize(text)
        tf = Counter(tokens)
        length = sum(tf.values())
        if length == 0:
            return
        vec = {t: float(c) for t, c in tf.items()}  # simple TF
        norm = math.sqrt(sum(v*v for v in vec.values()))
        self._docs[doc_id] = DocVector(doc_id, title, vec, norm, length)
        for t in tf.keys():
            self.doc_freq[t] += 1

    def finalize(self):
        self.N = len(self._docs)
        self.avgdl = sum(dv.length for dv in self._docs.values()) / max(1, self.N)

    def _vectorize_query(self, query: str) -> Tuple[Dict[str, float], float]:
        tf = Counter(tokenize(query))
        vec = {t: float(c) for t, c in tf.items()}
        norm = math.sqrt(sum(v*v for v in vec.values()))
        return vec, norm

    def search(self, query: str, top_k: int = 5) -> List[Tuple[DocVector, float]]:
        """Cosine similarity search over TF vectors (FAST first-stage retrieval)."""
        qvec, qnorm = self._vectorize_query(query)
        if qnorm == 0:
            return []
        results = []
        for dv in self._docs.values():
            dot = 0.0
            for t, qv in qvec.items():
                dvw = dv.vector.get(t)
                if dvw:
                    dot += qv * dvw
            score = dot / (qnorm * dv.norm) if dv.norm != 0 else 0.0
            results.append((dv, score))
        results.sort(key=lambda x: x[1], reverse=True)
        return results[:top_k]

# ----------------------------
# 3) Reranking with BM25 (lexical)
# ----------------------------

def bm25_score(query: str, dv: DocVector, doc_freq: Counter, N: int, avgdl: float, k1: float = 1.5, b: float = 0.75) -> float:
    """Classic BM25 lexical scoring for (query, document)."""
    q_terms = Counter(tokenize(query))
    score = 0.0
    for t, _ in q_terms.items():
        df = doc_freq.get(t, 0)
        if df == 0:
            continue
        idf = math.log((N - df + 0.5) / (df + 0.5) + 1.0)
        tf = dv.vector.get(t, 0.0)  # raw term count
        denom = tf + k1 * (1 - b + b * dv.length / (avgdl if avgdl > 0 else 1))
        score += idf * ((tf * (k1 + 1)) / (denom if denom > 0 else 1))
    return score

def rerank_bm25(query: str, candidates: List[Tuple[DocVector, float]], vdb: VectorDB, top_k: int = 5) -> List[Tuple[DocVector, float]]:
    """Rerank the initial candidates using BM25. Returns top_k results."""
    rescored = []
    for dv, _ in candidates:
        score = bm25_score(query, dv, vdb.doc_freq, vdb.N, vdb.avgdl)
        rescored.append((dv, score))
    rescored.sort(key=lambda x: x[1], reverse=True)
    return rescored[:top_k]

# ----------------------------
# 4) LLM "simulation" (a tiny answer composer)
# ----------------------------

def extractive_sentences(query: str, text: str, top_n: int = 3) -> List[str]:
    """Pick sentences that overlap most with the query terms (very simple heuristic)."""
    q = set(tokenize(query))
    sentences = re.split(r"(?<=[.!?])\s+", text.strip())
    scored = []
    for s in sentences:
        s_terms = set(tokenize(s))
        overlap = len(q & s_terms)
        scored.append((overlap, s))
    scored.sort(key=lambda x: x[0], reverse=True)
    return [s for _, s in scored[:top_n] if s]

def llm_generate_answer(query: str, top_chunks: List[DocVector]) -> Tuple[str, List[str]]:
    """
    Compose a small, readable answer that:
    - defines each concept found,
    - explains how they connect in a RAG pipeline,
    - and includes short extracts from sources.
    Returns (answer_text, citations)
    """
    bullets = []
    citations = []
    seen = set()
    for dv in top_chunks:
        if dv.doc_id in seen:
            continue
        seen.add(dv.doc_id)
        sample = extractive_sentences(query, DOC_TEXT[dv.doc_id], top_n=1)
        snippet = f" — e.g., \"{sample[0]}\"" if sample else ""
        bullets.append(f"- From **{dv.title}**{snippet}")
        citations.append(dv.title)

    intro = (
        "Here's a concise, step-by-step picture:\n\n"
        "• **LLM**: a generative model that turns a prompt into text.\n"
        "• **Vector DB**: stores embeddings to quickly find similar chunks.\n"
        "• **RAG**: retrieves relevant chunks, feeds them to the LLM for grounded answers.\n"
        "• **Reranking**: reorders retrieved chunks (e.g., with BM25) so the best evidence is on top.\n"
    )
    body = "\n".join(bullets) if bullets else "- (No supporting chunks found.)"
    outro = (
        "\n\nIn practice: embed your docs → store in a vector DB → retrieve top-k for a query → "
        "rerank those candidates → send the best few to the LLM as context → generate an answer (optionally with citations)."
    )
    return intro + body + outro, citations

# ----------------------------
# 5) Wiring: Build the store
# ----------------------------

DOC_TEXT = {d["id"]: d["text"] for d in DOCS}
DOC_TITLES = {d["id"]: d["title"] for d in DOCS}

def build_store() -> VectorDB:
    vdb = VectorDB()
    for d in DOCS:
        vdb.add(d["id"], d["title"], d["text"])
    vdb.finalize()
    return vdb

# ----------------------------
# 6) Pretty-print helpers
# ----------------------------

def print_header(title: str):
    print("\n" + "=" * 80)
    print(title)
    print("=" * 80)

def show_results(tag: str, results: List[Tuple[DocVector, float]]):
    print_header(tag)
    for rank, (dv, score) in enumerate(results, start=1):
        print(f"[{rank}] {dv.title:40s}  score={score:.4f}  (id={dv.doc_id})")

def run_demo(query: str, top_k: int = 4):
    vdb = build_store()

    print_header("Your query")
    print(query)

    # Stage 1: dense retrieval (cosine over TF vectors)
    initial = vdb.search(query, top_k=top_k)
    show_results("Stage 1 — Dense retrieval (cosine on TF vectors)", initial)

    # Stage 2: rerank with BM25 (lexical)
    reranked = rerank_bm25(query, initial, vdb, top_k=top_k)
    show_results("Stage 2 — Rerank (BM25)", reranked)

    # Compose an answer like a tiny RAG
    top_chunks = [dv for dv, _ in reranked[:3]]
    answer, citations = llm_generate_answer(query, top_chunks)

    print_header("RAG-style Answer (simulated LLM)")
    print(answer)
    print("\nSources:", ", ".join(citations) if citations else "(none)")

# ----------------------------
# 7) CLI
# ----------------------------

def main():
    parser = argparse.ArgumentParser(description="RAG Playground (LLM, Vector DB, Reranking).")
    parser.add_argument("--query", type=str, help="Your question, e.g., 'What is RAG and how does reranking help?'")
    parser.add_argument("--top_k", type=int, default=4, help="How many candidates to retrieve/rerank (default: 4).")
    args = parser.parse_args()

    if args.query:
        run_demo(args.query, top_k=args.top_k)
    else:
        try:
            q = input("Ask me something about LLMs, RAG, reranking, or vector DBs:\n> ").strip()
        except EOFError:
            q = "What is RAG and how does reranking help?"
        if not q:
            q = "What is RAG and how does reranking help?"
        run_demo(q, top_k=args.top_k)

if __name__ == "__main__":
    main()
PY
chmod +x /content/rag_demo.py
```
